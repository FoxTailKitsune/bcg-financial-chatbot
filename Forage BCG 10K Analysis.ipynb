```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\Jo\Documents\Forage BCG 10K Analysis.csv")
```

Profitability Metrics:

```python
df['Net Margin (%)'] = df['Net Income'] / df['Total Revenue'] * 100
df['ROA (%)'] = df['Net Income'] / df['Total Assets'] * 100
```

Leverage and Liquidity Ratios:

```python
df['Asset-to-Liability Ratio'] = df['Total Assets']/df['Total Liabilities']
df['Operating Cash Flow Ratio'] = df['Cash Flow from Operating Activities']/df['Total Liabilities']
```

Efficiency Metrics:

```python
df['Asset Turnover'] = df['Total Revenue'] / df['Total Assets']
df['Cash Flow Margin (%)'] = df['Cash Flow from Operating Activities']/df['Total Revenue'] * 100
```

Growth Measures:

```python
df['Revenue Growth (%)'] = df.groupby(['Company'])['Total Revenue'].pct_change() * 100
df['Net Income Growth (%)'] = df.groupby(['Company'])['Net Income'].pct_change() * 100
df['Operating Cash Flow Growth(%)'] = df.groupby(['Company'])['Cash Flow from Operating Activities'].pct_change() * 100
df['Margin Expansion (pp)'] = df.groupby('Company')['Net Margin (%)'].diff()
```

```python
rev_pivot_df = df.pivot(index = 'Company', columns = 'Year', values = 'Total Revenue')
rev_cagr = ((rev_pivot_df[2025] / rev_pivot_df[2023])**(1/2) - 1) * 100

inc_pivot_df = df.pivot(index = 'Company', columns = 'Year', values = 'Net Income')
inc_cagr = ((inc_pivot_df[2025] / inc_pivot_df[2023])**(1/2) - 1) * 100
```

```python
df['Revenue CAGR (%)'] = df['Company'].map(rev_cagr)
df['Net Income CAGR (%)'] = df['Company'].map(inc_cagr)

df = df.sort_values(by=['Company', 'Year'])
df
```

# BCG 10-K Financial Analysis Report (2023–2025)
## Performance Summary

```python
cagr_summary = pd.DataFrame({
    'Revenue CAGR %': rev_cagr,
    'Net Income CAGR %': inc_cagr
})

cagr_summary.round(2)
```

### Microsoft
* Not only does Microsoft have the highest revenue CAGR, but its net income CAGR shows *net income growth outpaced revenue growth*. This means that, over the past three years, Microsoft has become **more operationally efficient**, becoming more profitable for each dollar of revenue it earns.
### Apple
* Similarly, although at more modest values, Apple also continues to optimize its bottom line. Apple maintains its presence as a mature market leader.
### Tesla
* On the other hand, Tesla has faced a few difficult years. Revenue growth has flatlined, and *net income has halved* year over year. While Tesla has maintained sales volume, **increasing COGS or operating expenses** has been eating into its margins.

## Health and Solvency
### Asset-to-Liability Ratio

```python
solvency_trend = df.pivot(index = 'Year', columns = 'Company', values = 'Asset-to-Liability Ratio')
solvency_trend.round(2)
```

* Tesla maintained a very high (safe) ratio, with low debt compared to its assets. Microsoft has also been increasing its ratio, from 2.00 in 2023 up to 2.25 in 2025. On the other hand, Apple has much higher debt leverage, with a ratio much closer to 1. 

### Cash Flow Margin

```python
cfmargin_trend = df.pivot(index = 'Year', columns = 'Company', values = 'Cash Flow Margin (%)')
cfmargin_trend.round(2)
```

* Microsoft is incredibly good at converting revenue directly into operating cash flow, with nearly 50% cash flow margins. On the other hand, Tesla may struggle with liquidity.

## Conclusions
* Microsoft continues to be a **high-growth firm**, navigating its technological transitions incredibly well.
* Apple remains a solid organization capable of extracting value from a mature market.
* Tesla has struggled in the past few years. It *needs a new catalyst* to extract itself. **Monitor closely**.
